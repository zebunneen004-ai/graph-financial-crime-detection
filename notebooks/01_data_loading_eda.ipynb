```python
import sys
sys.path.append('src')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from load_data import load_elliptic_data, clean_and_merge, split_labeled_unlabeled

# Load data
features, classes, edges = load_elliptic_data('data')
df = clean_and_merge(features, classes)
df_all, df_labeled = split_labeled_unlabeled(df)

print(f"All nodes: {len(df_all)}")
print(f"Labeled nodes: {len(df_labeled)}")
print(f"Unknown nodes: {len(df_all) - len(df_labeled)}")
print(f"\nLabel distribution:")
print(df['label'].value_counts())
```

```python
# ============================================================
# CELL 2: BASIC STATISTICS & KEY INSIGHTS
# ============================================================

print("=" * 60)
print("DATASET OVERVIEW")
print("=" * 60)
print(f"Total transactions: {len(df_all):,}")
print(f"Labeled transactions: {len(df_labeled):,} ({len(df_labeled)/len(df_all)*100:.1f}%)")
print(f"Unknown transactions: {len(df_all) - len(df_labeled):,} ({(len(df_all)-len(df_labeled))/len(df_all)*100:.1f}%)")
print()

print("=" * 60)
print("CLASS DISTRIBUTION (Labeled Only)")
print("=" * 60)
illicit = (df_labeled['label'] == 1).sum()
licit = (df_labeled['label'] == 0).sum()
print(f"Licit (0): {licit:,} ({licit/len(df_labeled)*100:.1f}%)")
print(f"Illicit (1): {illicit:,} ({illicit/len(df_labeled)*100:.1f}%)")
print(f"Imbalance ratio: 1 illicit : {licit/illicit:.1f} licit")
print()

print("=" * 60)
print("TIME STEPS")
print("=" * 60)
print(f"Time steps range: {df_all['time_step'].min()} to {df_all['time_step'].max()}")
print(f"Total time steps: {df_all['time_step'].nunique()}")
print()

print("=" * 60)
print("CRITICAL AML INSIGHT")
print("=" * 60)
print(f"Only {len(df_labeled)/len(df_all)*100:.1f}% of transactions are labeled.")
print(f"Illicit class is {licit/illicit:.0f}x smaller than licit.")
print("This is why accuracy is misleading and PR-AUC/recall matter.")
```

```python
# ============================================================
# CELL 3: LABEL DISTRIBUTION VISUALIZATION
# ============================================================

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

# --- Plot 1: All labels (including unknown) ---
ax1 = axes[0]
labels_all = df['label'].value_counts()
colors_all = ['#3498db' if x == 0 else '#e74c3c' if x == 1 else '#95a5a6' 
              for x in labels_all.index]
bars1 = ax1.bar(labels_all.index.astype(str), labels_all.values, 
                color=colors_all, edgecolor='black', linewidth=0.5)

ax1.set_yscale('log')
ax1.set_xlabel('Label', fontsize=12)
ax1.set_ylabel('Count (Log Scale)', fontsize=12)
ax1.set_title('All Transactions: Label Distribution', fontsize=13, fontweight='bold')
ax1.set_xticks(['0', '1', 'unknown'])
ax1.set_xticklabels(['Licit (0)', 'Illicit (1)', 'Unknown'])

# Add value labels on bars
for bar in bars1:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height,
             f'{int(height):,}', ha='center', va='bottom', fontsize=10)

# --- Plot 2: Labeled only (linear scale) ---
ax2 = axes[1]
labels_labeled = df_labeled['label'].value_counts().sort_index()
colors_labeled = ['#3498db', '#e74c3c']
bars2 = ax2.bar(['Licit (0)', 'Illicit (1)'], labels_labeled.values, 
                color=colors_labeled, edgecolor='black', linewidth=0.5, width=0.6)

ax2.set_xlabel('Label', fontsize=12)
ax2.set_ylabel('Count', fontsize=12)
ax2.set_title('Labeled Transactions Only', fontsize=13, fontweight='bold')

# Add value labels with percentages
for bar in bars2:
    height = bar.get_height()
    pct = height / labels_labeled.sum() * 100
    ax2.text(bar.get_x() + bar.get_width()/2., height,
             f'{int(height):,}\n({pct:.1f}%)', 
             ha='center', va='bottom', fontsize=11)

plt.tight_layout()
plt.savefig('figures/label_distribution.png', dpi=300, bbox_inches='tight')
plt.show()
print("✅ Saved: figures/label_distribution.png")
```

```python
# ============================================================
# CELL 4: TEMPORAL ANALYSIS — TRANSACTIONS BY TIME STEP
# ============================================================

fig, axes = plt.subplots(2, 1, figsize=(14, 10))

# --- Plot 1: Transaction counts by time step ---
ax1 = axes[0]

# Count per time step
all_counts = df_all.groupby('time_step').size()
illicit_counts = df_labeled[df_labeled['label'] == 1].groupby('time_step').size()
licit_counts = df_labeled[df_labeled['label'] == 0].groupby('time_step').size()

# Reindex to include all 49 time steps
all_steps = range(1, 50)
illicit_counts = illicit_counts.reindex(all_steps, fill_value=0)
licit_counts = licit_counts.reindex(all_steps, fill_value=0)

ax1.bar(all_steps, licit_counts.values, color='#3498db', alpha=0.7, 
        label='Licit', edgecolor='none', width=0.8)
ax1.bar(all_steps, illicit_counts.values, bottom=licit_counts.values, 
        color='#e74c3c', alpha=0.9, label='Illicit', edgecolor='none', width=0.8)

ax1.set_xlabel('Time Step', fontsize=12)
ax1.set_ylabel('Number of Transactions', fontsize=12)
ax1.set_title('Labeled Transactions by Time Step (Stacked)', fontsize=13, fontweight='bold')
ax1.legend(fontsize=11, loc='upper right')
ax1.grid(True, alpha=0.3, axis='y')
ax1.set_xlim(0.5, 49.5)

# --- Plot 2: Illicit RATIO by time step (the key insight) ---
ax2 = axes[1]

total_labeled = licit_counts + illicit_counts
# Avoid division by zero
illicit_ratio = illicit_counts / total_labeled.replace(0, np.nan)

bars = ax2.bar(all_steps, illicit_ratio.values * 100, 
               color='#e74c3c', alpha=0.7, edgecolor='black', linewidth=0.3)

# Add horizontal line for overall average
avg_ratio = illicit_ratio.mean() * 100
ax2.axhline(y=avg_ratio, color='darkred', linestyle='--', linewidth=2, 
            label=f'Overall Average: {avg_ratio:.1f}%')

ax2.set_xlabel('Time Step', fontsize=12)
ax2.set_ylabel('Illicit Ratio (%)', fontsize=12)
ax2.set_title('Illicit Transaction Ratio by Time Step — KEY AML INSIGHT', 
              fontsize=13, fontweight='bold')
ax2.legend(fontsize=11)
ax2.grid(True, alpha=0.3, axis='y')
ax2.set_xlim(0.5, 49.5)
ax2.set_ylim(0, max(illicit_ratio.values * 100) * 1.2)

# Highlight the highest illicit ratio time step
max_step = illicit_ratio.idxmax()
max_ratio = illicit_ratio.max() * 100
ax2.annotate(f'Peak: {max_ratio:.1f}%\nat step {max_step}', 
             xy=(max_step, max_ratio), xytext=(max_step + 5, max_ratio + 5),
             arrowprops=dict(arrowstyle='->', color='darkred'),
             fontsize=10, fontweight='bold', color='darkred')

plt.tight_layout()
plt.savefig('figures/transactions_by_timestep.png', dpi=300, bbox_inches='tight')
plt.show()

print("✅ Saved: figures/transactions_by_timestep.png")
print(f"\n🔍 KEY FINDING:")
print(f"   Highest illicit ratio: {max_ratio:.1f}% at time step {max_step}")
print(f"   Lowest illicit ratio: {illicit_ratio.min()*100:.1f}% at time step {illicit_ratio.idxmin()}")
print(f"   This temporal clustering suggests coordinated illicit activity.")
```

```python
# ============================================================
# DATA QUALITY CHECK (Add to end of Notebook 1)
# ============================================================

print("=" * 60)
print("DATA QUALITY CHECK")
print("=" * 60)

# 1. Check for missing values in features
missing_features = df_all[[c for c in df_all.columns if 'feature_' in c]].isnull().sum().sum()
print(f"\n1. Missing values in feature columns: {missing_features}")
print("   ✅ Expected: 0 (Elliptic dataset is clean)")

# 2. Check for duplicate txIds
duplicates = df_all['txId'].duplicated().sum()
print(f"\n2. Duplicate transaction IDs: {duplicates}")
print("   ✅ Expected: 0")

# 3. Check edge validity (all txIds in edges exist in features)
edges_in_features = edges['txId1'].isin(df_all['txId']).all() and edges['txId2'].isin(df_all['txId']).all()
print(f"\n3. All edge nodes exist in features: {edges_in_features}")
print("   ✅ Expected: True")

# 4. Check time_step range consistency
time_range = df_all['time_step'].describe()
print(f"\n4. Time step statistics:")
print(f"   Min: {time_range['min']:.0f}, Max: {time_range['max']:.0f}")
print(f"   Mean: {time_range['mean']:.1f}, Std: {time_range['std']:.1f}")
print("   ✅ Expected: 1 to 49, roughly uniform")

# 5. Check label consistency
label_check = df_all['label'].value_counts()
print(f"\n5. Label consistency:")
print(f"   Unique labels: {df_all['label'].unique()}")
print(f"   Counts: {label_check.to_dict()}")
print("   ✅ Expected: 0, 1, 'unknown'")

# 6. Check for self-loops in edges (a transaction paying itself)
self_loops = (edges['txId1'] == edges['txId2']).sum()
print(f"\n6. Self-loops in graph: {self_loops}")
print("   ℹ️  Note: Bitcoin transactions CAN have change outputs to self")

print("\n" + "=" * 60)
print("DATA QUALITY: PASSED" if all([
    missing_features == 0,
    duplicates == 0,
    edges_in_features,
    self_loops >= 0  # Self-loops are possible
]) else "DATA QUALITY: REVIEW NEEDED")
print("=" * 60)
```
